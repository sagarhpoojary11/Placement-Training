Introduction to Matplotlib

```python
import matplotlib.pyplot as plt
import pandas as pd
```

Bar plot

```python
course_enroll={
    "courses":['java','python','c','c++'],
    "enrolled":[25,15,18,10]
}
course_df=pd.DataFrame(course_enroll)

#Create plot
plt.Figure(figsize=(9,5))
bars=plt.bar(course_df["courses"],course_df["enrolled"],
        color=["#07D100","#D1D100","#1504CD","#03011E"],
        edgecolor="black",width=0.6
)
#Creating Title and Label
plt.title("Course Enrolled Student Overview",fontsize=15,fontweight='bold',pad=5)
plt.xlabel("Courses",fontsize=12)
plt.ylabel("Course Enrolled Student",fontsize=12)
plt.bar_label(bars,padding=2,fontweight='bold',fontsize=12)
#Improve
plt.ylim(0,30)
plt.grid(axis='y',linestyle="--",alpha=0.3)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

Line plot

```python
weekly_sales={
    "days":['Mon','Tue','Wed','Thu','Fri','Sat','Sun'],
    "first_week":[35000,30000,38000,32000,39000,42000,45000],
    "second_week":[32000,35000,42000,37000,34000,43000,48000]
}
sales=pd.DataFrame(weekly_sales)
#sales=pd.data(figsize=(9,5))
#----------------first week sales------------
plt.plot(sales["days"],sales["first_week"],color="red",ls="--",marker="*",label="first week sales")
#----------------second week sales------------
plt.plot(sales["days"],sales["second_week"],color="blue",ls="--",marker="o",label="second week sales")

#create title 
plt.title("Shop weekly sales overview",fontweight="bold",fontsize=15)
plt.xlabel("Days of week",fontsize=12)
plt.ylabel("sales amount per day",fontsize=12)
plt.grid()
plt.legend()
plt.show()
```
