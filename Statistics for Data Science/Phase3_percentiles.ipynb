```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic_df=pd.read_csv('titanic.csv')
titanic_df.head(3)
```

```python
#Data
titanic_df.isnull().sum()
```

```python
#Replace NaN value with mean 
meanAge=titanic_df['Age'].mean()
titanic_df['Age']=titanic_df['Age'].fillna(meanAge)
```

```python
#For age Find Percentiles
print("0%(min):",np.percentile(titanic_df["Age"],0))
print("25%:",np.percentile(titanic_df["Age"],25))
print("50%(median):",np.percentile(titanic_df["Age"],50))
print("75%:",np.percentile(titanic_df["Age"],75))
print("100%(max):",np.percentile(titanic_df["Age"],100))


```

```python
sns.boxplot(x='Age',data=titanic_df)
plt.show()
```

```python
#for fare find percentiles
print("0%(min):",np.percentile(titanic_df["Fare"],0))
print("25%:",np.percentile(titanic_df["Fare"],25))
print("50%(median):",np.percentile(titanic_df["Fare"],50))
print("75%:",np.percentile(titanic_df["Fare"],75))
print("100%(max):",np.percentile(titanic_df["Fare"],100))
```

```python
sns.boxplot(x='Fare',data=titanic_df)
plt.show()
```
