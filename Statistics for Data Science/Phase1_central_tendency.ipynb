Measure of Central Tendency

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load dataset
titanicdf=pd.read_csv('titanic.csv')
titanicdf.head(10)
```

Mean

```python
meanVal=np.mean(titanicdf['Fare'])
meanVal
```

```python
sns.histplot(x=titanicdf['Fare'],data=titanicdf,bins=[i for i in range(0,151,10)])
plt.plot([meanVal for i in range(0,351)],[i for i in range(0,351)],c='red')
plt.ylim(0,400)
plt.show()
```

Meadian

```python
medianVal=np.median(titanicdf['Fare'])
medianVal
```

```python
sns.histplot(x=titanicdf['Fare'],data=titanicdf,bins=[i for i in range(0,151,10)])
plt.plot([medianVal for i in range(0,351)],[i for i in range(0,351)],c='yellow')
plt.ylim(0,400)
plt.show()
```

Mode

```python
import statistics as st
modeVal=st.mode(titanicdf['Fare'])
modeVal
```

```python
sns.histplot(x=titanicdf['Fare'],data=titanicdf,bins=[i for i in range(0,151,10)])
plt.plot([modeVal for i in range(0,351)],[i for i in range(0,351)],c='orange')
plt.ylim(0,400)
plt.show()
```

Final Ouput


```python
sns.histplot(x=titanicdf['Fare'],data=titanicdf,bins=[i for i in range(0,151,10)])
plt.plot([meanVal for i in range(0,351)],[i for i in range(0,351)],c='red',label="mean")
plt.plot([medianVal for i in range(0,351)],[i for i in range(0,351)],c='yellow',label="median")
plt.plot([modeVal for i in range(0,351)],[i for i in range(0,351)],c='orange',label="mode")
plt.legend()
plt.ylim(0,400)
plt.title("Titanic Ticket Fare Central Tendency",fontweight='bold',fontsize=13)
plt.tight_layout()
plt.show()
```
