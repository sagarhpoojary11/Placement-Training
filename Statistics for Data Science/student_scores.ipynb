```python
import pandas as pd
```

```python
df=pd.read_csv("student_scores.csv")
```

```python
df["Average_Score"]=df[["MathScore","ReadingScore","WritingScore"]].mean(axis=1)
```

```python
def calculate_mad(x):
    mean=x.mean()
    return(x-mean).abs().mean()

result=df.groupby("EthnicGroup")["Average_Score"].agg(
    Highest="max",
    Minimum="min",
    Mean="mean",
    MAD=calculate_mad
)

print(result.round(2))
```

```python
overall_mean=df["Average_Score"].mean()
```

```python
result["Difference"] = abs(result["Mean"] - overall_mean)
closest = result["Difference"].idxmin()

print("Group closest to overall mean:", closest)
print("Group Mean:", result.loc[closest, "Mean"])
```
