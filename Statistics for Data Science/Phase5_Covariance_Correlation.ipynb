Covariance and Correlation

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales_df=pd.read_csv('monthly_sales.csv')
sales_df.head(10)
```

```python
#Filter January Month data from sales_df
sales_Jan=sales_df[sales_df['Sales Month']=='Jan']
sales_Jan.tail(2)
#Find the cov,corr between Advertising V/S Sales
#sales_Jan['Advertising'].cov(sales_Jan['Sales'])
sales_Jan['Advertising'].corr(sales_Jan['Sales'])
```

```python
sns.scatterplot(data=sales_Jan,x='Advertising',y='Sales')
plt.title("Correlation (Jan)Advertising vs Sales")
plt.show()
```
