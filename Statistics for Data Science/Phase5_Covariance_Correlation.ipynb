### Covariance and Correlation

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales_df=pd.read_csv('monthly_sales.csv')
sales_df.head(2)
```

### No Correlation

#### Filter January Month Data from sales_df

```python
sales_Jan=sales_df[sales_df['Sales Month']=='Jan']
sales_Jan.tail(10)
# Find the cov, cor between Advertising V/S Sales
# cov = sales_Jan['Advertising'].cov(sales_Jan['Sales'])
# corr=sales_Jan['Advertising'].corr(sales_Jan['Sales'])
# print(corr)
```

```python
sns.scatterplot(data=sales_Jan,x='Advertising',y='Sales')
plt.title("Correlation (Jan) Advertising V/S Sales")
plt.show()
```

```python
#Filter February Month data from sales_df
sales_Feb=sales_df[sales_df["Sales Month"]=='Feb']
#find the corr between Advertising V/S Sales
corr = sales_Feb["Advertising"].corr(sales_Feb["Sales"])
print(corr)
```

```python
sns.scatterplot(data=sales_Feb,x='Advertising',y='Sales')
plt.title("Correlation (Feb) Advertising V/S Sales")
plt.show()
```

```python
#Filterr March Month data from sales_df
sales_Mar=sales_df[sales_df['Sales Month']=='Mar']
#find the corr between Advertising V/S Sales
corr = sales_Mar["Advertising"].corr(sales_Mar["Sales"])
print(corr)
```

```python
sns.scatterplot(data=sales_Mar,x='Advertising',y='Sales')
plt.title("Correlation (Mar) Advertising V/S Sales")
plt.show()
```

### #Find Corr --> tips Dataset

```python
tips_df.isnull().sum()
tips_df.info()
```
