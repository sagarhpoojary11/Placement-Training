```python
import pandas as pd
```

Rows&Col - selections

```python
sdata=pd.read_csv("subdata_pandas/student1.csv")
sdata
```

```python
#To select columns
sdata[['firstName','fees','age']]
# To select details of karthik
sdata.loc[sdata.firstName=="Karthik"]
sdata.loc[sdata.fees>=30000]
#to slicing on rows
sdata.loc[1:7]
# to slicing on rows & columns
sdata.iloc[:4,:3]
sdata.iloc[2:7,1:4]
```

Filter

```python
sdata=pd.read_csv("subdata_pandas/student1.csv")
sdata
```

```python
sdata['age']>20
sdata[sdata['age']>20]
student_age_filter=sdata[sdata['age']>20]
student_age_filter
```

```python
sdata.where(sdata['age']>20)
sdata.where(sdata['age']>20,other="Minor")
```

```python
sdata.loc[(sdata['marks']>=85,"result")]="selected"
sdata.loc[(sdata['marks']<85,"result")]="not selected"
sdata
```
