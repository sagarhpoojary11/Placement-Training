Dict Functions & Methods

```python
student={
    'id':"4MW24CS119",
    'name':"Sagar",
    'age':20,
    'dept':"CSEB"
}
#Retrive a single value using key
get_value=student.get('age')
print(get_value)
#check all the keys 
dict_keys=student.keys()
print(dict_keys)
#Retrieve all the values
dict_values=student.values()
print(dict_values)
#Update 
update_age=student.update({"age":21})
print(student)
#set default
dep_val=student.setdefault("dept","CS")
print(dep_val)
new_loc=student.setdefault("place","Udupi")
print(student)
#copy
c=student.copy()
print(c)
#pop - remove key-value based on key
student.pop("dept")
print(student)
#popitem
student.popitem()
print(student)
#dict_items
print(student.items())
```

Dict iteration

```python
student={
    'id':"4MW24CS119",
    'name':"Sagar",
    'age':20,
    'dept':"CSEB"
}
#Itr over keys
for k in student:
    print(k)
#Itr over values
for v in student.values():
    print(v)
#Itr over key-values 
for k,v in student.items():
    print(k,":",v)
```

write a script to generate a lookup dictionary where key represent risk tiers(1 to 15) and values are the aquare of each tier number.

```python
dict_square={}
for i in range(1, 16):
    dict_square.setdefault(i,i**2)
print(dict_square)
```

you manage a retail sales register where customer checkout totals arrive throughout the day in a dictionary format :{"customer Name":amount_paid}. write a python program to sort this sales dictionary by transaction amount in descending order and identify the top-paying customer.

```python
sales = {
    "Rahul": 2500,
    "Sagar": 4500,
    "Priya": 3200,
    "Amit": 1800
}

# Sort by amount in descending order
sorted_sales = dict(sorted(sales.items(), key=lambda x: x[1], reverse=True))

print("Sales in descending order:")
print(sorted_sales)

# Find top-paying customer
top_customer = max(sales, key=sales.get)

print("Top-paying customer:", top_customer)
print("Amount paid:", sales[top_customer])
```
