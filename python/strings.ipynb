String Indexing & Slicing

```python
str = "DARKCHOCOLATE"
print("org:",str)
#Idx +ve
print("5th",str[5])#retrive the 5th element
print("last",str[len(str)-1])#retrive the last element
#Idx -ve
print(str[-1])#print the last element
print(str[-len(str)])#print the first element
```

```python
#Slicing - String_name[start:stop:step]
#Default                 0  len(s)  1
#print(str[0:4]) #Dark step=1
#print(str[:4]) #Dark start=0 step=1
#print(str[4:len(str)]) #CHOCOLATE stop=len(str) step=1
#print(str[0:len(str):2]) #DRCOOAE
print(str[12:3:-1]) #ETALOCOHC
#rev_string
rev_string = str[::-1]
print(rev_string)
print(str[4:9])
print(str[9:])
```

String Functions

```python
s = "raju told me that he is a coder"
print(s+"It's real")#concetnation
print(len(s)) #length
print(s.endswith("der"))# is string endswith the substring
print(s.replace("he is","I am"))#replace old to new
print(s.find("a")) #first index of substring
print(s.count("e")) #Count the freq of substring
print(s.capitalize()) #capitalize the first letter of string
print(s.upper()) #upper case
print(s.lower()) #lower case
```
