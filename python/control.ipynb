While loop

```python
num=3876
count=0
while num>0:
    num=num//5
    count+=1
print(count)
```

```python
n=int(input("Enter the first number"))
m=int(input("Enter the second number"))
for i in range(1,m+1):
    print(f"{i}*{n}={i*n}")#print(i,'*',n,"=",i*n)
```
