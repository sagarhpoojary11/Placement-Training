Tuples Immutable

```python
colors=("red","blue","green")
#         0     1       2
#colors[1]="pink" #does not support
print(colors)
#If you want change convert to list & change
list_colors=list(colors)
list_colors[1]="pink"
colors=tuple(list_colors)
print(colors,type(colors))
```

Tuple Packing & Unpacking

```python
v1="Raj"
v2=23
v3="CSE"
#packing
student=v1,v2,v3
#unpacking
name,age,dept=student
print(name)
print(age)
print(dept)
```

Tuple Methods & Functions

```python
nums=(10,30,50,20,40,15)
print(max(nums))#Max
print(min(nums))#Min
sort_nums=sorted(nums) #Sorting
print(sort_nums)
print("count",nums.count(50))
print("Index",nums.index(30))
```
