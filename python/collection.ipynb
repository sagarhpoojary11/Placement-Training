```python
#------List Functions------
colors=["red","green","black","blue"]
print("org List:",colors)
print(len(colors)) #Length
colors.append("gold") #Insert new element end of the list
print("After adding Gold",colors)
colors[1]="white" #replace an element
print("After replace",colors)
colors.insert(2,"pink") #Insert value in a Idx position
print("After Insert pink 2idx",colors) 
colors.remove("black")#remove an element
print("After removing black",colors)
colors.pop()
print("After popping",colors)
```

```python
#Looping
colors=["red","green","black","blue"]
#Using index Access the list element
for i in range(len(colors)):
    print(i,colors[i])
#Directly -> Access the list element
for c in colors:
    print(c,end=" ")
```

```python
marks_sem1=[34,54,21,45,65,76]
print("Max: ",max(marks_sem1))
print("Min: ",min(marks_sem1))
marks_sem1.sort()
print("Marks after sorting:",marks_sem1)
marks_sem1.reverse()
print("Marks after reversing:",marks_sem1)
```

```python
marks_sem1=[34,45,67,32,87,90]
marks_sem2=[134,145,165,190,132,187]
#final_marks=marks_sem1+marks_sem2
#print(final_marks)
#store both sem marks in marks_sem1 list
marks_sem1.extend(marks_sem2)
print(marks_sem1)
```

```python
#List Comprehensions
#rule -> new_list = [expression for item in iterable if condition]
n=10
even =[i for i in range(1,n+1) if i%2==0]
print(even)

x=5
sqr=[x**2 for x in range(1,x+1)]
print(sqr)
```
