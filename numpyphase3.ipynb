Numpy Functions

```python
import numpy as np
```

Aggregate Function

```python
arr=np.array([10,20,30,40,50])
print("Sum:",np.sum(arr))
print("Max:",np.max(arr))
print("Min:",np.min(arr))
print("Count:",arr.size)
print("Cumulitive sum:",np.cumsum(arr))
print("Cumulitive product:",np.cumprod(arr))
```

Statistical Function

```python
import statistics as st
sales=np.array([200,180,150,250,270,300,250,230,500,1000])
print("Mean:",np.mean(sales))
print("Median:",np.median(sales))
print("Mode:",st.mode(sales))
print("Standard Deviation:",np.std(sales))
```

```python
arr=np.array([19,23,17,22,18])
print(np.std(arr))
arr1=np.array([15,20,30,10,50])
print(np.std(arr1))
```

```python
price=np.array([120,180,200,210,220,300])
sales=np.array([15,13,10,12,9,5])
print(np.corrcoef([price,sales]))
```

```python
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

# find the sales of data
# print sample data for first three restaurent
# Total sale per year
# Maximum sales per a student
# Maximum sales every year
# Average sales of every restaurant
# Average cummulative sales occurs all the restaurant
# Monthly Average for all the restaurant every year
sales = sales_data[:, 1:]

print("Sales Data:")
print(sales)

print("\nFirst three restaurants:")
print(sales[:3])

print("\nTotal sales per year:")
print(np.sum(sales, axis=0))

print("\nMaximum sales per restaurant:")
print(np.max(sales, axis=1))

print("\nMaximum sales every year:")
print(np.max(sales, axis=0))

print("\nAverage sales of every restaurant:")
print(np.mean(sales, axis=1))

print("\nCumulative sales:")
print(np.cumsum(sales, axis=0))

print("\nAverage sales for all restaurants every year:")
print(np.mean(sales, axis=0))
```
